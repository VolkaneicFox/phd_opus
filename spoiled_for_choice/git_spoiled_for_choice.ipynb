    <div style="margin: 0;"><b>Abstract.</b> <i> Nothing is true... </i> Multi-model prediction tensors present the problem of model-choice. This paper circumvents this problem with a localised precision-weighting ensemble framework that aggregates signals across diverse inductive biases and cross-validation folds to generate robust point-predictions. The final point estimator predicts yields for rice and wheat farms with competitive accuracies, $R^2=0.54$ and $R^2=0.59$, respectively. 

<i> Everything is permitted... </i> This paper prioritizes robustness over model- or fold-level outperformance due to the heterogeneity inherent in Indian agriculture. The candidate models with higher accuracy have higher weight in the ensemble estimator, simultaneously, models are penalized if they over-fit. Empirical results suggest that the given set of models -- Random Forest, XGBoost and Support Vector Regression -- have homoscedastic prediction risk, consequently the arithmetic mean is the robust estimator. 

A ``sanity-check'', based on Stability Ratio, ensures that farm-level predictions are dominated by between-model diversity rather than within-model sampling noise; 98\% farms qualify this reliability threshold.

The prediction residuals are heteroscedastic, as seen in the U-shaped error distribution at yield tails. This is exploited to identify farms with high tail-risk, using conditional quantile prediction intervals. Applying dispersion weighting to this subsample of high prediction uncertainty farms, improves point-prediction $R^2$ by 10\%. Finally, Shapley-based feature importance is used to highlight the relative importance of climate conditions, input costs and farm-management. Together, this creates a comprehensive farm-level profile that integrates robust median forecasts with explicit measures of risk exposure and feature influence, that can potentially inform agricultural policy and risk-management.</div>
