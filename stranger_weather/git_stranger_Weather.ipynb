<div style="font-size: 2.2em; line-height: 1.3;font-family: 'Sitka Banner', 'Sitka Text', Georgia, serif; text-align:center; width:100%;">
<div> <u>Stranger Weather: Estimating the Growth Shock from Historic and Structural Deviations in the Local Weather of 400 German Counties, 1995–2019
</u> </div>

<div style="font-size: 1.5em; line-height: 1.3;font-family: 'Sitka Banner', 'Sitka Text', Georgia, serif; text-align:center; width:100%;">
    Oliver Holtemöller and Sukanya Mukherjee</div>
<div style="font-size: 0.75em; line-height: 1.3;font-family: 'Sitka Banner', 'Sitka Text', Georgia, serif; text-align:center; width:100%;">
    IWH Halle (Saale) and MLU Halle (Saale)</div>

<div style="font-size: 1.1em; line-height: 1.4; font-family: 'Sitka Banner', 'Sitka Text', Georgia, serif; text-align: justify; width: 85%; margin: 0 auto; padding: 15px 20px; border: 1px solid #ccc; background-color: #fafafa; border-radius: 4px;"><b>Abstract.</b> Semantics underneath the definitions of weather anomalies can be exploited to answer questions such as: which regions are facing accelerated warming, is the weather deviating more from its seasonal or monthly expectations, and, importantly, <i>does it matter?</i> This paper, based on evidence from four hundred German counties between 1995 and 2019, finds that these distinctions are non-trivial. Regional warming above respective upper thresholds has not been uniform across Germany; by updating these thresholds annually, this paper finds that southern counties are warming faster than the north. This accelerated warming, above seasonal moving thresholds, is associated with a 0.11 percentage point reduction in the growth rate of GDP per-capita. Alternatively, above monthly moving thresholds, it is associated with 0.12-0.24 percentage points reduction; the upper bound follows when the share of days with conservative temperatures are included in the average annual warming. On annual fluctuations in temperature, the seasonal mean temperatures deviate roughly 0.03&degC more than monthly mean temperatures around their respective annual means. Subsequently, the growth loss associated with a 1 standard deviation increase in within-year variability is 0.50 (seasonal) and 0.33 (monthly) percentage points. The question - "does it matter" - remains central to the interpretations presented in this paper. In the end, this paper attempts to contribute to existing climate discourse, the assessment that how weather or climate anomalies are defined, is critical for measuring climate risk, identifying geographical vulnerabilities and estimating potential economic losses, with the goal of informing diverse economic decisions.</div>

<div style="font-size: 1.25em; line-height: 1.3; font-family: 'Sitka Banner', 'Sitka Text', Georgia, serif; text-align: justify; width: 100%;">
  <p style="text-align: justify; text-justify: inter-word; margin-bottom: 12px;"><b>"Strange weather"</b> is defined as divergence between present weather conditions and its expectations. The expected weather in a given month or season is derived from its relevant historical distribution, extensively discussed in Vogel et al. [2020], Waidelich et al. [2024] and Kotz et al. [2024]. Current weather is considered "strange" or "unexpected" if it falls outside of 90% confidence of what it was supposed to be, this is addressed in this paper as the 5th and 95th percentile tails of the reference distribution. That is, a high temperature or precipitation anomaly is identified as weather conditions falling in the right-tail of the respective distribution, while a low temperature or precipitaiton anomaly is identified as weather conditions falling in the left-tail of the respective distribution. </p>
<p style="text-align: justify; text-justify: inter-word; margin-bottom: 12px;">A second source of "strange weather" can be ascribed to intra-annual fluctuations, that is, second order effects in line with Burke et al. [2015], Kotz et al. [2021], Kalkuhl et al. [2020] and Linsenmeier [2024]. These are deviations from the county expectations of the contemporaneous yearly climate distribution. If the expectations on contemporary climatic conditions are the sequence of monthly and seasonal means of temperature and precipitation, then fluctuations from this sequence determine the intra-annual "strange weather." This paper uses intra-annual standard deviation over monthly means and seasonal means, as the measure of variability. To address potential noise-induced biases in the coefficients, tail events are explicitly removed from the means. The "strangeness" in a year, therefore, is ascribed to how far months/seasons deviate from each other. </p>
<p style="text-align: justify; text-justify: inter-word; margin-bottom: 12px;"><b>Contributions.</b> This paper contributes to the climate-growth literature in three ways. First, it treats extremes and fluctuations as separable, empirically distinguishable channels, rather than collapsing "strange weather" into a single composite measure. Second, it contributes to the discussion on how much of climate's economic impact is geographic versus income- and development-driven. Essentially, by focusing on a wealthy, cold economy, this paper shows that measurable negative growth effects from weather fluctuations and excess warming persist even where poverty and low development cannot be the explanation. Third, it demonstrates that within-country heterogeneity in baseline climatic conditions is itself a driver of weather-shock impact — geography shapes vulnerability even holding income and institutions constant, making it a non-trivial factor in discussions of climate risk.
</p>
<p style="text-align: justify; text-justify: inter-word; margin-bottom: 0;">Identification: Two-way fixed effects (county and year) absorb time-invariant county characteristics and common annual shocks, ensuring the fixed effects framework isolates genuine local anomalies rather than persistent county-level patterns. County-specific quadratic trends further net out slow-moving confounders like demographic shift and capital accumulation. What remains, ceteris paribus, is variation in local weather that is "strange" relative to otherwise identical counties in the same year — isolating the growth effect of weather surprise itself, net of climate, trend, and common shock.</p>
</div>

<p style="text-align: center;"><u>Surprise Heat (°C) above 95th Percentile Thresholds of County Distributions of Daily Temperatures</u></p></div>

<!-- Body Text Container -->
<div style="font-size: 1.25em; line-height: 1.3; font-family: 'Sitka Banner', 'Sitka Text', Georgia, serif; width: 100%;">
  <p style="text-align: justify; text-justify: inter-word; margin-bottom: 12px;">Divergence in the local temperature conditions for German counties is measured against the distribution of daily  mean temperatures in 1950-1980, hereafter "fixed", or against thirty-year "moving" history. For 2015, for example, the reference distribution spans 1980-2010, for 2016, 1981-2011, and so on. Since daily annual temperatures are higher during summer months and lower during winter months, the distributions are separated by seasons and by months and the 5th and 95th percentile thresholds are identified from these bands. The annual excess warming or excess cooling (°C) is the average of seasonal and monthly deviations from the thresholds. </p> 
 <p style="text-align: justify; text-justify: inter-word; margin-bottom: 12px;"><u>Seasonally Distinct Distributions of Daily Mean Temperatures:</u> "Surprise Heat (°C)" above the 95th percentile threshold of the thirty-year moving historical distributions has a robust negative impact on the annual growth rates of German counties, during 1995-2019. <b>1 standard deviation increase in surprise heat above the moving threshold is associated with 0.11 percentage points reduction in the annual growth-rate of county GDP per-capita </b>. This effect is tempered when "null surprises", that is days with 0°C excess warming, are included in the yearly average, and is significant only when the tail events are excluded from the intra-annual variability in temperature.  </p>
<p style="text-align: justify; text-justify: inter-word; margin-bottom: 0;"><u> Monthly Distinct Distributions of Daily Mean Temperatures:</u> "Surprise Heat (°C)" measured against monthly thresholds behave differently to that measured against seasonal thresholds, mainly that growth loss is largest when "null surprises" are included and high temperature events are identified using the fixed 1950-1980 monthly distribution. <b> A 1 standard deviation more warming above the 95th percentile monthly threshold (1950-80 reference) is associated with 0.24 percentage points decline in the annual country GDP per-capita growth rate. For moving thresholds, the effect is around 0.13 percentage points </b>.</p>

<p style="text-align: center;"><u>Intra-Annual Variability in Seasonal and Monthly Mean Temperatures (°C)</u></p>
</div>

<!-- Body Text Container -->
<div style="font-size: 1.25em; line-height: 1.3; font-family: 'Sitka Banner', 'Sitka Text', Georgia, serif; width: 100%;">
  <p style="text-align: justify; text-justify: inter-word; margin-bottom: 12px;">The intra-annual variability in the temperature across German counties is measured as the standard deviation of seasonal and monthly mean temperatures. The clean measure excludes the days within each season or month that have temperatures below the 5th or above the 95th percentile thresholds of the relevant reference distribution. Alternatively, these tail events are also included in the measure, but the coefficients estimated in the regressions are robust to this specification.</p>
<p style="text-align: justify; text-justify: inter-word; margin-bottom: 12px;">The coefficient of intra-annual variability in <u>seasonal mean temperatures</u> is most negative when tail events are excluded from the calculation of the seasonal mean temperature, and when the reference distribution is fixed. In this case, 1 standard deviation increase in the intra-annual variability is associated with roughly <b>0.80 percentage points reduction in the growth rate</b>. When the reference is the thirty-year moving distribution, the effect is reduced and 1 standard deviation increase corresponds to roughly <b>0.60 percentage points reduction in the growth rate</b>. Including the days with mean temperatures in the tails of the seasonal distributions leads to an attenuation bias since the noise is, possibly, moderating the true effect of the fluctuations in mean temperatures in the year.</p>
<p style="text-align: justify; text-justify: inter-word; margin-bottom: 0;">The coefficient of intra-annual variability in <u>monthly mean temperatures</u> is most negative when tail events are excluded from the calculation of the monthly means, and when the reference distribution is fixed to 1950-80. This is similar to the pattern observed with seasonal mean temperatures. However, in the case of monthly means, including the tail events makes the coefficients of intra-annual variability, statistically insignificant. Therefore, the impact is restricted to cases when the variability is measured without monthly "noise". 1 standard deviation increase in the variability of monthly mean temperatures is associated with, roughly, <b>0.51 percentage points reduction in the annual growth rate of county GDP per-capita</b>. The effect is lower when the reference distribution is "moving", and <b>1 standard deviation increase in variability corresponds to 0.35 percentage points reduction in the growth rate</b>.</p>

    <p style="text-align: justify; text-justify: inter-word; margin: 0;"><b>Fig 2a. Correlation between County Growth-Rate and Intra-Annual Variability in Seasonal Mean Temperatures (°C): </b> The correlation is prominently negative along the northern coastline and into the northwest — precisely where intra-annual temperature variability is structurally lowest in Germany. This pattern speaks directly to an open question in the temperature-variability literature: whether the elevated growth sensitivity observed in equatorial and lower-income economies reflects income itself, or a geographic artifact whereby <u>regions with low endemic variability are least equipped to absorb variability when it does arrive</u>. The German sub-national panel offers a rare within-country test of this hypothesis, holding income, institutions, and development largely constant. In the moderate-to-high variability counties of central and southwestern Germany, the correlation weakens and, in several cases, turns positive — consistent with an adaptation story rather than a purely income-driven one. Cross-sectional heterogeneity aside, the two-way fixed effects panel estimation confirms a robust, statistically significant negative effect of intra-annual variability on annual growth nationally — geography may shape <i>who</i> is exposed, but it does not change the sign of the underlying cost.</p>

This means July and September's own hot days are judged against a threshold already inflated by August's aggressive warming, and are correspondingly less likely to register as extreme relative to that pooled bar. The seasonal measure therefore concentrates its signal on the season's worst-performing month, while systematically understating warming in its neighboring months. Under the monthly measure, July, August, and September are each judged against their own distribution — so July and September's warming shows up as real excess heat in their own right, even when milder than August's. The result is more days flagged as extreme events under the monthly measure: roughly three additional flagged days per year under fixed thresholds, and roughly one under moving thresholds, relative to the seasonal count — but a smaller average excess-heat magnitude, since the measure is no longer borrowing August's severity to judge its neighbors. In short: <b>the seasonal measure trades event count for event intensity, concentrating the signal into fewer, hotter breaches; the monthly measure spreads the same underlying warming across more, individually milder ones, and ends up with more days flagged as "extremes".</b>  </div>

<div style="font-size: 1.8em; line-height: 1.3;font-family: 'Sitka Banner', 'Sitka Text', Georgia, serif; text-align:center; width:100%;">
<div> <u>Conclusion</u> </div>

    
<div style="font-size: 0.75em; line-height: 1.3;font-family: 'Sitka Banner', 'Sitka Text', Georgia, serif; text-align:justify; width:100%;">  
<div>Climate fluctuations disrupt both biological (Harrison et al. [2006]) and economic (Kotz et al. [2021]) systems, and anthropogenic warming continues to destabilize the climatic patterns (Zhou et al. [2023]). This paper contributes to that literature by identifying local weather fluctuations as deviations from historical expectation, and by showing that the definitional choice underlying that comparison — present seasons against their past, or present months against theirs — is not a technicality, but changes the estimated cost of warming itself. A 1 standard deviation warming (°C) above the seasonal upper threshold corresponds to roughly a 0.10 percentage point reduction in county GDP per-capita growth, fairly consistent across specifications. Against the monthly upper threshold, the effect is larger and less stable across specifications: 0.12 to 0.24 percentage points. A related difference emerges when daily temperatures are smoothed to seasonal or monthly averages: intra-annual temperature variability over the seasonal series costs 0.34–0.82 percentage points of growth, while over the monthly series, the cost is slightly smaller, at 0.30–0.60 percentage points. </div>

<div style="font-size: 1em; line-height: 1.3;font-family: 'Sitka Banner', 'Sitka Text', Georgia, serif; text-align:justify; width:100%;">
<div>For intra-annual temperature variability, smoothing daily temperatures to monthly averages without excluding days outside the 90% inter-percentile band causes the coefficient to lose significance entirely. For seasons, the coefficients grow substantially larger once extreme days are excluded from the smoothing. The decision to clean the smoothing is therefore not incidental: it directly resolves an attenuation bias that would otherwise dilute the robustness of intra-annual temperature fluctuations (Linsenmeier [2024]). </div>
    
<div style="font-size: 1em; line-height: 1.3;font-family: 'Sitka Banner', 'Sitka Text', Georgia, serif; text-align:justify; width:100%;">
<div>On surprise heat (°C) above the 95th-percentile threshold, this paper further examines the choice between historical references — a distant, fixed anchor (1950–1980) versus a more contemporaneous expectation, a thirty-year moving reference ending five years prior to the present. The moving threshold is motivated by the need to capture accelerating warming, and this paper finds a distinct upward trend in that acceleration, exploiting its cross-sectional heterogeneity across German counties to estimate growth loss. For monthly extremes, the fixed reference distribution yields the more robust coefficients; for seasonal extremes, the moving reference is more robust. This asymmetry underscores that the reference window is itself a modeling choice with real consequences for estimated impact, not a robustness afterthought.
</div>
<div style="font-size: 1em; line-height: 1.3;font-family: 'Sitka Banner', 'Sitka Text', Georgia, serif; text-align:justify; width:100%;">
<div> In conclusion, building on the extensive discourse surrounding the economic impact of climate variability, this paper finds that deviations in local weather conditions — whether from historical expectation, or through fluctuation within the year — correspond to significant losses in the growth rate of GDP. Additionally, by focusing on a wealthy, cold economy, this paper shows that geography — not merely income or development — can account for meaningful variation in climate vulnerability. Counties in the north, which have historically experienced the lowest intra-annual temperature variability, show the strongest negative correlation between that variability and growth: the regions least accustomed to fluctuation appear least equipped to absorb it. Future research should further interrogate how the choice of reference period itself shapes perceived climate risk — since what counts as "normal" is, itself, a moving target.</div>
